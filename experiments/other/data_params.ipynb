# Noise term (sigma)
sigma is a parameter that controls the standard deviation of the noise added to the outcome variable (y).

The default value from the `synthetic_data` function is 1.0.

Our experiments are run with sigma=3.0.

```python
exp = Experiment("sigma", sigma=[0.5, 1,  2,  3, 4,  5], topk_methods=main_no_random)
exp.plots(hue="topK_algorithm")
```

# Number of covariates
p (int optional): number of covariates (>=5)

The default value is 10.

```python
exp = Experiment("p", p=[5, 10,20], topk_methods=main_no_random)
exp.plots(hue="topK_algorithm")
```

# Modes for the synthetic data generation process

mode (int, optional): mode of the simulation: 
1. for difficult nuisance components and an easy treatment effect. 
2. for a randomized trial. 
3. for an easy propensity and a difficult baseline. 
4. for unrelated treatment and control groups. 
5. for a hidden confounder biasing treatment.

```python
#    N_FEATURES = 5
#    SIGMA = 1

exp = Experiment("mode", mode=[1,2,4,5], topk_methods=main_no_random)
exp.plots(hue="topK_algorithm")
```
