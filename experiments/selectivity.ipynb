```python
%load_ext autoreload
%autoreload 2

from trees.data import Data
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


N = 10000

allss = []

for i in range(30):
    data = Data()
    data.generate(n=N, seed=i)

    _ , s = data.generate_random_condition(max_s=1)
    allss.append({"max_s": 1, "s":s})

    _ , s = data.generate_random_condition(max_s=0.95)
    allss.append({"max_s": 0.95, "s":s})

df = pd.DataFrame(allss)
```

```python
sns.histplot(df[df["max_s"] == 1]["s"])
plt.title("max_s = 1")
```

```python
sns.histplot(df[df["max_s"] == 0.95]["s"])
plt.title("max_s = 0.95")
```
