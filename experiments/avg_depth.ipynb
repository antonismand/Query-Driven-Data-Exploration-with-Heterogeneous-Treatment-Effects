```python
%load_ext autoreload
%autoreload 2

from trees.causal_tree import CTP
from trees.topk import *
from trees.data import Data
import statistics

for i in range(10):
    D = Data()
    D.generate(seed=i)

    alg = CTP()
    alg.fit(D=D)
    D.generate_random_condition()
    alg.online()
    depths = [x['depth'] for i, x in alg.subgroups.items()]
    
    scan = Random(alg)
    top = scan.get_topK()
    print(round(statistics.mean(depths),1), top['depth'].mean())
```
