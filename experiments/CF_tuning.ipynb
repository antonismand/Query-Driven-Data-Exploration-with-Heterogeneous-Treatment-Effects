```python
from econml.dml import CausalForestDML
from tqdm import tqdm 
from trees.data import Data


for i in tqdm(range(10)):
    data = Data()
    data.generate(n=10000, seed=i)

    cf = CausalForestDML(
                discrete_treatment=True,
                random_state=123,
                n_estimators = 100,
            )

    before_tuning = cf.__dict__.copy()

    cf.tune(
            X=data.df[data.feature_names].to_numpy(),
            Y=data.df["outcome"].to_numpy(),
            T=data.df["treatment"].to_numpy(),
    )

    # print only the differences between the two dicts
    for k,v in cf.__dict__.items():
        if before_tuning[k] != v:
            print(f"{k}: {before_tuning[k]} -> {v}")


    cf.fit(
            X=data.df[data.feature_names].to_numpy(),
            Y=data.df["outcome"].to_numpy(),
            T=data.df["treatment"].to_numpy(),
    )

    print(cf.models_t)
    print(cf.models_y)
```
