```python
from trees.causal_forest import CFT
from trees.topk import *
from trees.data import Data
from loguru import logger

for mode in ["UTC", "UTC-f1-f4", "UTC-f0-f9"]:
    logger.info(f"------------ mode={mode} ------------" )
    D = Data()
    D.generate(seed=0, mode=mode)

    alg = CFT()
    start = time()
    alg.fit(D=D)
    
    alg.get_important_features(0)
```

```python
from trees.experiment import Experiment

exp = Experiment("mode", exp_name="feature_importances", mode=["UTC", "UTC-f1-f4", "UTC-f0-f9"])
```

```python
exp.plots(x="Top-K Algorithm", hue="mode")
```
